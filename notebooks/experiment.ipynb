* DF: jumlah dokumen yang mengandung suatu kata $t$
> DF('apel') = 2 dokumen

* IDF

$$
IDF(t) = \log \left(\frac{N}{DF}\right) = \log \left(\frac{\text{jumlah dokumen}}{DF}\right)
$$

## Term Frequency

# _for hiding_

```python
def termFrequency(document, vocab):
    tf = dict()
    count = 0
    for token in document:
        if token in vocab:
            count += 1
            freq = {token: count}
            tf.update(freq)
            count = 0
    return tf
```

```python
tf_list = []
for docs in cleaned_documents:
    docs = termFrequency(docs, vocab)
    tf_list.append(docs)
```

```python
def docFrequency(TF):
    docList = []
    docFreq = dict()
    for termfreqs in TF:
        for keys in termfreqs:
            docList.append(keys)
    
    for term in docList:
        freq = {term: docList.count(term)}
        docFreq.update(freq)
        
    return docFreq
```

### IDF

$$
IDF(t) = \log \left(\frac{N}{DF(t)}\right) = \log \left(\frac{\text{jumlah dokumen}}{DF(t)}\right)
$$

```python
idf = calcIDF(texts.shape[0], docFreq)
idf
```

```python
# idf
import numpy as np
N = texts.shape[0] # banyaknya dokumen
idf = {}
for df in docFreq:
    idf.update({df: np.log(N/docFreq.get(df))})
```

## TF-IDF

```python
tf_idf = calcTFIDF(termFreq, docFreq, idf)
tf_idf
```

```python
tf_idf = [] 

for tf in termFreq:
    tf_idf_doc = {}
    for tokens in tf:
        tf_idf_doc[tokens] = docFreq[tokens] * idf[tokens]
    tf_idf.append(tf_idf_doc)
```
