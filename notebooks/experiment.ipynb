# **tf-idf sample**

* dokumen 1: 'apel merah enak'
* dokumen 2: 'apel hijau segar'
* dokumen 3: 'apel merah segar'

* DF: jumlah dokumen yang mengandung suatu kata $t$
> DF('apel') = 2 dokumen

* IDF

$$
IDF(t) = \log \left(\frac{N}{DF}\right) = \log \left(\frac{\text{jumlah dokumen}}{DF}\right)
$$

## Term Frequency
Jumlah kemunculan kata dalam dokumen

* dokumen 1: 'apel' = 1, 'merah' = 1, 'enak' = 1
* dokumen 2: 'apel' = 1, 'hijau' = 1, 'segar' = 1
* dokumen 3: 'apel' = 1, 'merah' = 1, 'segar' = 1

# _for hiding_

```python
def termFrequency(document, vocab):
    tf = dict()
    count = 0
    for token in document:
        if token in vocab:
            count += 1
            freq = {token: count}
            tf.update(freq)
            count = 0
    return tf
```

```python
tf_list = []
for docs in cleaned_documents:
    docs = termFrequency(docs, vocab)
    tf_list.append(docs)
```

```python
def docFrequency(TF):
    docList = []
    docFreq = dict()
    for termfreqs in TF:
        for keys in termfreqs:
            docList.append(keys)
    
    for term in docList:
        freq = {term: docList.count(term)}
        docFreq.update(freq)
        
    return docFreq
```

```python
def calcTermFrequency(documents):
    termFreqList = []

    for document in documents:
        tf_doc = {} # term frequency for every documents
        for token in document:
            tf_doc[token] = tf_doc.get(token, 0) + 1
        
        termFreqList.append(tf_doc)
    return termFreqList

def calcDocFrequency(termFreqList):
    docFreq = []
    docFreqList = {}
    for tf_doc in termFreqList:
        for keys in tf_doc:
            docFreq.append(keys)

    for tokens in docFreq:
        freq = {tokens: docFreq.count(tokens)}
        docFreqList.update(freq)
        
    return docFreqList
```

```python
termFreq = calcTermFrequency(texts)
```
