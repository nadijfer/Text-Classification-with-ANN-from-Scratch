# **tf-idf sample**

* dokumen 1: 'apel merah enak'
* dokumen 2: 'apel hijau segar'
* dokumen 3: 'apel merah segar'

* TF: jumlah kemunculan kata dalam dokumen
> dokumen 1: 'apel' = 1, 'merah' = 1
> dokumen 2: 'apel' = 1, 'hijau' = 1

* DF: jumlah dokumen yang mengandung suatu kata $t$
> DF('apel') = 2 dokumen

* IDF

$$
IDF(t) = \log \left(\frac{N}{DF}\right) = \log \left(\frac{\text{jumlah dokumen}}{DF}\right)
$$

## Term Frequency

```python
def termFrequency(document, vocab):
    tf = dict()
    count = 0
    for token in document:
        if token in vocab:
            count += 1
            freq = {token: count}
            tf.update(freq)
            count = 0
    return tf
```

```python
tf_list = []
for docs in cleaned_documents:
    docs = termFrequency(docs, vocab)
    tf_list.append(docs)
```

```python
tf_list
```

```python
def docFrequency(TF):
    docList = []
    docFreq = dict()
    for termfreqs in TF:
        for keys in termfreqs:
            docList.append(keys)
    
    for term in docList:
        freq = {term: docList.count(term)}
        docFreq.update(freq)
        
    return docFreq
```

```python
for i in range(len(texts)):
    texts.iloc[i] = preprotext(texts.iloc[i])
```

```python
def calcTermFrequency(documents, vocab):
    termFreq = dict()
    for document in documents:
        count = 0
        for token in document:
            if token in vocab:
                count += 1
                freq = {token: count}
                termFreq.update(freq)
    return termFreq

def calcDocFrequency(termFreqList):
    docList = []
    docFreq = dict()
    for keys in termFreqList:
        docList.append(keys)
    
    for term in docList:
        freq = {term: docList.count(term)}
        docFreq.update(freq)
        
    return docFreq
```

```python
termFreq = calcTermFrequency(texts, vocab)
termFreq
```

```python
docFreq = calcDocFrequency(termFreq)
docFreq
```
